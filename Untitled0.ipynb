```python

import random
import streamlit as st

st.title("Guess the Number Game")
st.write("I have chosen a number between 1 and 100. Can you guess it?")

if "number" not in st.session_state:
    st.session_state.number = random.randint(1, 100)
    st.session_state.attempts = 0
    st.session_state.game_over = False

guess = st.number_input(
    "Enter your guess",
    min_value=1,
    max_value=100,
    step=1
)

if st.button("Guess"):
    if st.session_state.game_over == False:
        st.session_state.attempts = st.session_state.attempts + 1

        if guess == st.session_state.number:
            st.success("Congratulations! You guessed the correct number!")
            st.write("Total attempts:", st.session_state.attempts)
            st.session_state.game_over = True

        elif guess > st.session_state.number:
            st.warning("Too high! Try again.")

        else:
            st.info("Too low! Try again.")

if st.button("Play Again"):
    st.session_state.number = random.randint(1, 100)
    st.session_state.attempts = 0
    st.session_state.game_over = False
    st.rerun()
```
